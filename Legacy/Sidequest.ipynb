```python
import json
from astroquery.gaia import Gaia
from pathlib import Path

CREDS_PATH = Path("gaia_credentials.json")

with CREDS_PATH.open("r", encoding="utf-8") as f:
    creds = json.load(f)

Gaia.login(user=creds["user"], password=creds["password"])
print("Gaia login OK")
```

```python
import run_stars

run_stars.run(
    mode=1,
    csv_path="Targets.csv",
    hostnames=["TOI-2431"],
    vlim=3.0,
    srad=90.0,
    verbose=True,
    attempt_timeout_s=1200,
    max_attempts=1,
    all_runs_dir="ALL_RUNS(Sidequest)",
    all_csvs_dir="ALL_CSVS(Sidequest)",
    force_rerun=True,
)
```

```python
import Comove

outdir = Comove.findfriends(
    "TOI-2431",
    12.09,
    velocity_limit=5.0,
    search_radius=100.0,
    radec=[37.763667, 8.381994],
    output_directory=None,
    verbose=True,
    showplots=False,
)
print("Returned:", outdir)
```
