```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "browser"

# ============================================
# USER INPUTS
# ============================================
TARGETS_CSV = "Targets.csv"
SUMMARY_CSV = "FRIENDS/friends_density_summary.csv"

TARGET_NAME = "HIP_67522"
NEIGHBOR_CSV = f"FRIENDS/{TARGET_NAME}.csv"

HOSTNAMES_TO_CHECK = [
]

# If True, also highlight every target from Targets.csv that appears in this sphere
HIGHLIGHT_ALL_TARGETS_IN_SPHERE = True

# ============================================
# COLUMN DEFINITIONS
# ============================================
N_RA_COL = "RA"
N_DEC_COL = "DEC"
N_PLX_COL = "Plx(mas)"
POSSIBLE_NEIGHBOR_ID_COLS = ["GaiaDR3", "Source", "source_id", "DR3Name"]

POSSIBLE_NAME_COLS = ["hostname", "target", "Target", "pl_name", "name"]
T_RA_COL = "ra"
T_DEC_COL = "dec"
T_PLX_COL = "st_parallax_mas"
T_GAIA_ID_COL = "gaia_dr3_id"

SUMMARY_NAME_COL = "target"
SUMMARY_RAW_COL = "raw_score"
SUMMARY_ADJ_COL = "adjusted_score"

# ============================================
# HELPERS
# ============================================
def sph_to_cart(ra_deg, dec_deg, plx_mas):
    d_pc = 1000.0 / plx_mas
    ra = np.deg2rad(ra_deg)
    dec = np.deg2rad(dec_deg)

    x = d_pc * np.cos(dec) * np.cos(ra)
    y = d_pc * np.cos(dec) * np.sin(ra)
    z = d_pc * np.sin(dec)
    return x, y, z, d_pc

def normalize_name(x):
    return str(x).strip().lower().replace("_", " ")

def parse_gaia_id(x):
    """
    Robust Gaia ID parser.
    Handles ints, floats, scientific notation, strings.
    Returns normalized string ID or pd.NA
    """
    if pd.isna(x):
        return pd.NA

    s = str(x).strip()
    if s == "" or s.lower() in {"nan", "none", "<na>"}:
        return pd.NA

    try:
        return str(int(float(s)))
    except Exception:
        digits = "".join(ch for ch in s if ch.isdigit())
        return digits if digits else pd.NA

# ============================================
# LOAD TARGETS
# ============================================
targets = pd.read_csv(TARGETS_CSV)

name_col = next((c for c in POSSIBLE_NAME_COLS if c in targets.columns), None)
if name_col is None:
    raise ValueError(
        f"No valid name column found in Targets.csv. "
        f"Tried {POSSIBLE_NAME_COLS}. "
        f"Columns are: {targets.columns.tolist()}"
    )

required_target_cols = [T_RA_COL, T_DEC_COL, T_PLX_COL, T_GAIA_ID_COL]
for col in required_target_cols:
    if col not in targets.columns:
        raise ValueError(f"Missing required target column: {col}")

targets = targets.copy()
targets["_norm"] = targets[name_col].map(normalize_name)
targets["_gid"] = targets[T_GAIA_ID_COL].map(parse_gaia_id)

# ============================================
# BUILD MASTER TARGET LOOKUP
# ============================================
all_host_to_id = {}
all_id_to_host = {}

for _, row in targets.iterrows():
    nm = row["_norm"]
    gid = row["_gid"]

    if pd.notna(nm) and pd.notna(gid):
        if nm not in all_host_to_id:
            all_host_to_id[nm] = gid
        if gid not in all_id_to_host:
            all_id_to_host[gid] = nm

all_target_id_set = set(all_id_to_host.keys())

# ============================================
# LOAD SUMMARY / SCORE LOOKUP
# ============================================
summary = pd.read_csv(SUMMARY_CSV)

required_summary_cols = [SUMMARY_NAME_COL, SUMMARY_RAW_COL, SUMMARY_ADJ_COL]
for col in required_summary_cols:
    if col not in summary.columns:
        raise ValueError(f"Missing required summary column: {col}")

summary = summary.copy()
summary["_norm"] = summary[SUMMARY_NAME_COL].map(normalize_name)

score_lookup = {}
for _, row in summary.iterrows():
    nm = row["_norm"]
    raw = row[SUMMARY_RAW_COL]
    adj = row[SUMMARY_ADJ_COL]

    if nm not in score_lookup:
        score_lookup[nm] = {
            "raw_score": raw,
            "adjusted_score": adj
        }

# ============================================
# LOAD MAIN TARGET
# ============================================
target_matches = targets[targets["_norm"] == normalize_name(TARGET_NAME)]

if len(target_matches) == 0:
    raise ValueError(f"Target '{TARGET_NAME}' not found in Targets.csv")

target_row = target_matches.iloc[0]

tx, ty, tz, _ = sph_to_cart(
    float(target_row[T_RA_COL]),
    float(target_row[T_DEC_COL]),
    float(target_row[T_PLX_COL])
)

target_gid = target_row["_gid"]
target_norm = normalize_name(TARGET_NAME)
target_scores = score_lookup.get(
    target_norm,
    {"raw_score": np.nan, "adjusted_score": np.nan}
)

# ============================================
# LOAD NEIGHBORS
# ============================================
df = pd.read_csv(NEIGHBOR_CSV)

neighbor_id_col = next((c for c in POSSIBLE_NEIGHBOR_ID_COLS if c in df.columns), None)
if neighbor_id_col is None:
    raise ValueError(
        f"No Gaia ID column found in neighbor CSV. "
        f"Tried {POSSIBLE_NEIGHBOR_ID_COLS}. "
        f"Columns are: {df.columns.tolist()}"
    )

df = df[(df[N_PLX_COL] > 0) & pd.notna(df[N_PLX_COL])].copy()
df["_gid"] = df[neighbor_id_col].map(parse_gaia_id)

x, y, z, d = sph_to_cart(
    df[N_RA_COL].astype(float).values,
    df[N_DEC_COL].astype(float).values,
    df[N_PLX_COL].astype(float).values
)

df["x_pc"] = x - tx
df["y_pc"] = y - ty
df["z_pc"] = z - tz
df["r_pc"] = np.sqrt(df["x_pc"]**2 + df["y_pc"]**2 + df["z_pc"]**2)

# ============================================
# MATCH ONLY REQUESTED HOSTNAMES
# ============================================
check_names_norm = [normalize_name(x) for x in HOSTNAMES_TO_CHECK]

host_to_id = {}
for nm in check_names_norm:
    gid = all_host_to_id.get(nm, pd.NA)
    if pd.notna(gid):
        host_to_id[nm] = gid

id_to_host = {v: k for k, v in host_to_id.items()}

matched_requested = df[df["_gid"].isin(set(host_to_id.values()))].copy()
matched_requested["matched_hostname"] = matched_requested["_gid"].map(id_to_host)

matched_requested["raw_score"] = matched_requested["matched_hostname"].map(
    lambda x: score_lookup.get(x, {}).get("raw_score", np.nan)
)
matched_requested["adjusted_score"] = matched_requested["matched_hostname"].map(
    lambda x: score_lookup.get(x, {}).get("adjusted_score", np.nan)
)

print("\n=== Main target score ===")
print(
    pd.DataFrame([{
        "target": TARGET_NAME,
        "raw_score": target_scores["raw_score"],
        "adjusted_score": target_scores["adjusted_score"]
    }]).to_string(index=False)
)

print("\n=== Requested hostnames ===")
for h in HOSTNAMES_TO_CHECK:
    print("-", h)

print("\n=== Requested hostnames found inside this sphere ===")
if len(matched_requested) == 0:
    print("None")
else:
    print(
        matched_requested[
            ["matched_hostname", "_gid", "r_pc", "raw_score", "adjusted_score"]
        ]
        .sort_values("r_pc")
        .to_string(index=False)
    )

# ============================================
# MATCH ALL TARGETS IN THIS SPHERE
# ============================================
matched_all_targets = df[df["_gid"].isin(all_target_id_set)].copy()
matched_all_targets["matched_hostname"] = matched_all_targets["_gid"].map(all_id_to_host)

# remove the central target itself if it appears
matched_all_targets = matched_all_targets[
    matched_all_targets["_gid"] != target_gid
].copy()

matched_all_targets["raw_score"] = matched_all_targets["matched_hostname"].map(
    lambda x: score_lookup.get(x, {}).get("raw_score", np.nan)
)
matched_all_targets["adjusted_score"] = matched_all_targets["matched_hostname"].map(
    lambda x: score_lookup.get(x, {}).get("adjusted_score", np.nan)
)

print("\n=== Any target from Targets.csv found inside this sphere ===")
if len(matched_all_targets) == 0:
    print("None")
else:
    print(
        matched_all_targets[
            ["matched_hostname", "_gid", "r_pc", "raw_score", "adjusted_score"]
        ]
        .sort_values("r_pc")
        .to_string(index=False)
    )

# ============================================
# PLOT
# ============================================
fig = go.Figure()

# Main cloud
fig.add_trace(go.Scatter3d(
    x=df["x_pc"],
    y=df["y_pc"],
    z=df["z_pc"],
    mode="markers",
    marker=dict(
        size=2.5,
        color=df["r_pc"],
        colorscale="Viridis",
        opacity=0.85,
        showscale=True,
        colorbar=dict(title="r (pc)")
    ),
    name="Neighbors",
    hoverinfo="skip"
))

# Highlight only explicitly requested hosts in red
if len(matched_requested) > 0:
    fig.add_trace(go.Scatter3d(
        x=matched_requested["x_pc"],
        y=matched_requested["y_pc"],
        z=matched_requested["z_pc"],
        mode="markers+text",
        marker=dict(
            size=6,
            color="red"
        ),
        text=matched_requested["matched_hostname"],
        textposition="top center",
        name="Requested matched hosts",
        hoverinfo="skip"
    ))

# Highlight all targets from Targets.csv found in this sphere
if HIGHLIGHT_ALL_TARGETS_IN_SPHERE and len(matched_all_targets) > 0:
    fig.add_trace(go.Scatter3d(
        x=matched_all_targets["x_pc"],
        y=matched_all_targets["y_pc"],
        z=matched_all_targets["z_pc"],
        mode="markers+text",
        marker=dict(
            size=5,
            color="red",
            symbol="diamond"
        ),
        text=matched_all_targets["matched_hostname"],
        textposition="top center",
        name="All target-list matches",
        hoverinfo="skip"
    ))

# Central target
fig.add_trace(go.Scatter3d(
    x=[0],
    y=[0],
    z=[0],
    mode="markers+text",
    marker=dict(
        size=7,
        color="red"
    ),
    text=[TARGET_NAME],
    textposition="top center",
    name="Target",
    hoverinfo="skip"
))

fig.update_layout(
    title=f"3D Neighbor Distribution centered on {TARGET_NAME}",
    scene=dict(
        xaxis_title="X (pc)",
        yaxis_title="Y (pc)",
        zaxis_title="Z (pc)",
        aspectmode="data"
    ),
    margin=dict(l=0, r=0, b=0, t=40)
)

fig.show()
```
