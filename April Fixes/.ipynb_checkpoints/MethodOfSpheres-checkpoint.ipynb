```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.cm import get_cmap


def normalize_name(s):
    return str(s).strip().replace(" ", "_")


# --- Reload everything fresh ---
summary_df = pd.read_csv(r"FRIENDS\friends_density_summary.csv")
confirmed_df = pd.read_csv(r"confirmed_clusters.csv")

summary_df.columns = summary_df.columns.str.strip()
confirmed_df.columns = confirmed_df.columns.str.strip()

summary_df["target"] = summary_df["target"].apply(normalize_name)
confirmed_df["target"] = confirmed_df["target"].apply(normalize_name)

if "cluster" not in confirmed_df.columns:
    raise ValueError("confirmed_clusters.csv must contain a 'cluster' column")

if "notes" not in confirmed_df.columns:
    confirmed_df["notes"] = ""

confirmed_df["cluster"] = confirmed_df["cluster"].fillna("").astype(str).str.strip()
confirmed_df["notes"] = confirmed_df["notes"].fillna("").astype(str).str.strip()

# --- Merge now, no reliance on old summary_df state ---
df = summary_df.merge(
    confirmed_df[["target", "cluster", "notes"]],
    on="target",
    how="left"
)

df["cluster"] = df["cluster"].fillna("").astype(str).str.strip()
df["is_confirmed_cluster"] = df["cluster"] != ""

# --- Keep only plottable rows ---
df = df[df["score_status"] == "ok"].copy()
df = df[np.isfinite(df["raw_score"]) & np.isfinite(df["score_retention"])].copy()

print("Rows available for plotting:", len(df))
print("Confirmed matched:", int(df["is_confirmed_cluster"].sum()))

matched = df.loc[df["is_confirmed_cluster"], ["target", "cluster"]].sort_values(["cluster", "target"])
print("\nMatched confirmed targets:")
print(matched.to_string(index=False) if not matched.empty else "None")

friends_targets = set(df["target"])
confirmed_targets = set(confirmed_df["target"])
unmatched_confirmed = sorted(confirmed_targets - friends_targets)

print("\nConfirmed targets NOT present in FRIENDS summary:")
print(unmatched_confirmed if unmatched_confirmed else "None")

# --- Plot ---
plt.figure(figsize=(9, 7))

# Base layer
base = df.loc[~df["is_confirmed_cluster"]]
plt.scatter(
    base["raw_score"],
    base["score_retention"],
    s=35,
    alpha=0.75,
    color="steelblue",
    label="Targets"
)

# Confirmed clusters by cluster color
confirmed = df.loc[df["is_confirmed_cluster"]].copy()
cluster_names = sorted([c for c in confirmed["cluster"].unique() if c != ""])

if cluster_names:
    cmap = get_cmap("tab10", len(cluster_names))
    for i, cluster_name in enumerate(cluster_names):
        sub = confirmed[confirmed["cluster"] == cluster_name]
        plt.scatter(
            sub["raw_score"],
            sub["score_retention"],
            s=90,
            color=cmap(i),
            edgecolors="black",
            linewidths=0.8,
            label=cluster_name
        )
else:
    print("\nNo confirmed clusters were matched into the plot.")

plt.axvline(0.6, linestyle="--", linewidth=2, color="black")
plt.axhline(0.8, linestyle="--", linewidth=2, color="black")

plt.xlim(0.2, 1.02)
plt.ylim(0.0, 1.02)

plt.xlabel("Raw score")
plt.ylabel("Adjusted / Raw")
plt.title("Cluster structure vs sampling reliability")
plt.legend()
plt.show()
```
