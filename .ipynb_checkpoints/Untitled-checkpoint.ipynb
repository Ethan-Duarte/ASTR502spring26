```python
import json
from astroquery.gaia import Gaia
from pathlib import Path

CREDS_PATH = Path("gaia_credentials.json")

with CREDS_PATH.open("r", encoding="utf-8") as f:
    creds = json.load(f)

Gaia.login(user=creds["user"], password=creds["password"])
print("Gaia login OK")
```
