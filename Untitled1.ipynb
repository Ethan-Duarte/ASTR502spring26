```python
import numpy as np

# Read the CSV with named columns
data = np.genfromtxt("Targets.csv", delimiter=",", names=True, dtype=None, encoding=None)

# Get min/max for ra and dec
ra_min = np.min(data["ra"])
ra_max = np.max(data["ra"])
dec_min = np.min(data["dec"])
dec_max = np.max(data["dec"])

print(f"ra:  min = {ra_min}, max = {ra_max}")
print(f"dec: min = {dec_min}, max = {dec_max}")
```
